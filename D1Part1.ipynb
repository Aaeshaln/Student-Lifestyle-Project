### Script for Descriptive Analysis

```python
import pandas as pd

# Read the dataset
df = pd.read_csv("student_lifestyle_dataset.csv")


# Descriptive statistics function
def Desc_stat(ds, var):

    summary = {
        "Count": ds[var].count(),
        "Mean": ds[var].mean(),
        "Median": ds[var].median(),
        "Mode": ds[var].mode(),
        "Minimum": ds[var].min(),
        "Maximum": ds[var].max(),
        "Range": ds[var].max() - ds[var].min(),
        "Standard Deviation": ds[var].std(),
        "Variance": ds[var].var(),
        "Skewness": ds[var].skew(),
        "Kurtosis": ds[var].kurt(),
        "25th Percentile": ds[var].quantile(0.25),
        "50th Percentile": ds[var].quantile(0.50),
        "75th Percentile": ds[var].quantile(0.75),
        "Interquartile Range": (
            ds[var].quantile(0.75)
            - ds[var].quantile(0.25)
        )
    }

    return summary


# Apply the function to GPA
result = Desc_stat(df, "GPA")

for key, value in result.items():
    print(f"{key:<25}: {value}")
```

### Random Sampling

```python
random_sample = df.sample(
    n=150,
    replace=False,
    random_state=42
)

# Apply the descriptive statistics function
random_result = Desc_stat(random_sample, "GPA")

print("Statistic\tGPA")

for key, value in random_result.items():
    if key == "Mode":
        continue
    print(f"{key}\t{value:.4f}" if key != "Count" else f"{key}\t{value}")
```
